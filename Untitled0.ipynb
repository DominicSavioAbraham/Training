```python
n = int(input("no of monkeys: "))
k = int(input("no of banana eaten by monkeys: "))
j = int(input("no of peanuts eaten by monkeys: "))
m = int(input("no of total banana: "))
p = int(input("no of total peanut: "))

if n <= 0 or k <= 0 or j <= 0 or m < 0 or p < 0:
    print("Invalid Input")
else:
    tb = m // k
    tp = p // j

    monkeys = tb + tp

    if m % k != 0 or p % j != 0:
        monkeys += 1

    if monkeys > n:
        monkeys = n

    print("Number of Monkeys left on the tree:", n - monkeys)
```
